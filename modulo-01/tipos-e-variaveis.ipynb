## Tipagem Dinâmica

A tipagem dinâmica no Python significa que você não precisa declarar o tipo de uma variável ao criá-la; o interpretador descobre o _tipo_ automaticamente em tempo de execução com base no valor atribuído.

Além disso, a capacidade de reatribuir valores/tipos (int, bool, str) totalmente diferentes a uma mesma variável é uma outra característica da tipagem dinâmica.

```python
# teste o código a baixo

x = 10 

print(x)
```

```python
# Agora altere o valor de x e imprima o novo resultado

x = 10

print (x)
```

## Tipos dos dados

Com a função type( ) podemos ter o retorno do tipo da variável que passarmos como parâmetro. 

```python
idade, nome, altura, adulta = 18, "Ana", 1.70, True

print(type(idade))
print(type(nome)) 
print(type(altura))
print(type(adulta))
```

```python
# Agora testem: 

x = 25
y = "25"

print(type(x))
print(type(y))
```

## Concatenação das Informações

Concatenar nada mais é do que juntar informações. 

Em Python é possível fazer concatenção de informações com: 

- Vírgulas
- f-string
- com o símbolo **+**


#### Com vírgulas:

```python
nome = "Ana"
idade = 14

print("A", nome, "tem", idade, "anos.")
```

#### Com f-string:

```python
nome = "Ana"
idade = 14

print(f"A {nome} tem {idade} anos.")
```

#### Com + (mais):

```python
nome = "Ana"
idade = 14

print("A " + nome + " tem " + str(idade) + " anos.")
```
