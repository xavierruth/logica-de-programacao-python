# Concatenando valores booleanos

É póssivel formatar valores booleanos para que ele saia de uma forma mais legível e entendível? Sim! 

Através do uso de operadores ternários ou condicionais simples, por meio da formatação utilizando o f-string.

#### Utilizando operador ternário dentro da f-string - _mais indicado_

Execute o código abaixo e veja como é a saída dessas informações.

```python
status = True

# Convertendo para Sim ou Não
print(f"O usuário está ativo? { 'Sim' if status else 'Não' }")

# Convertendo para Ligado ou Desligado
print(f"Dispositivo: { 'Ligado' if status else 'Desligado' }")
```

E podemos até mesmo incrementar a saída com emojis...

```python
sucesso = True
erro = False

print(f"Status do upload: { '✅ Sucesso' if sucesso else '❌ Falha' }")
print(f"Conexão com o banco: { '🟢 Online' if not erro else '🔴 Offline' }")
```

### E qual a diferença das aspas duplas ("") para as aspas simples ('') nesse caso? 

Apenas pelo fato de não causar confusão! Como o f-string já inicia com as _aspas duplas_, o texto colocando dentro das chaves { } precisam utilizar as _aspas simples_, evitando que o python se confunda sobre onde o texto começa e termina.
