# Expressões e Operações em Python

### 1. Operadores Aritméticos

Utilizados para cálculos matemáticos básicos e avançados.

```python
a = 5 
b = 2
```

```python
# Adição (+)
print("Adição (+):", a + b)
# 5 + 2 = 7
```

```python
# Subtração (-)
print("Subtração (-):", a - b)
# 5 - 2 = 7
```

```python
# Multiplicação (*)
print("Multiplicação (*):", a * b)
# 5 * 2 = 10
```

```python
# Divisão (/)
print("Divisão (/):", a / b)
# 5 / 2 = 2.5 
```

```python
# Módulo (%)
print("Módulo (%):", a % b)
# 5 % 2 = 1 (útil para saber se é par ou ímpar)
```

```python
# Exponenciação (**)
print("Exponenciação (**):", a ** b)
# 5 elevado a 2 = 25 -> 5 * 5 = 25 
```

#### 1.1 Ordem de Prioridade (Precedência)
A hierarquia de execução das operações matemáticas.

```python
resultado1 = 10 + 5 * 2
print("10 + 5 * 2 =", resultado1)    
# 10 + 10 = 20
```

```python
print(10 + 5 * 2)
```

```python
resultado2 = (10 + 5) * 2
print("(10 + 5) * 2 =", resultado2)  
# 15 * 2 = 30
```

```python
print((10 + 5)*2)
```

### 2. Operadores de atribuição

Servem para definir ou atualizar o valor armazenado em uma variável.

```python
x = 10
print("Atribuição simples (=):", x) 
print(x)
```

```python
x += 3   # Equivalente a: x = x + 3
print("Adição e atribuição (+=):", x)
print(x)
```

```python
x -= 2   # Equivalente a: x = x - 2
print("Subtração e atribuição (-=):", x)
print(x)
```

```python
x *= 2   # Equivalente a: x = x * 2
print("Multiplicação e atribuição (*=):", x)
print(x)
```

```python
x /= 4   # Equivalente a: x = x / 4
print("Divisão e atribuição (/=):", x)
print(x)
```

```python
x %= 3   # Equivalente a: x = x % 3
print("Módulo e atribuição (%=):", x)
print(x)
```

### 3. Operadores Relacionais/Comparação

Comparam dois valores e retornam sempre um valor booleano (True ou False).

```python
n1 = 1 
n2 = 9
```

```python
# comparação de igualdade (==)
print("Igual n1 (==):", n1 == n2)
```

```python
# comparação de diferença (!=)
print("Diferente de (!=):", n1 != n2)
```

```python
# comparação de maior que (>)
print("Maior que (>):", n1 > n2)
```

```python
# comparação de menor que (<)
print("Menor que (<):", n1 < n2)
```

```python
# comparação de maior ou igual a (>=)
print("Maior ou igual n1 (>=):", n1 >= 9)
```

```python
# comparação de menor ou igual a (<=)
print("Menor ou igual n1 (<=):", n1 <= n2)
```

### 4. Operadores lógicos

Utilizados para conectar múltiplas expressões condicionais.

```python
idade = 20 
possui_cnh = True
dia = "Sábado"
```

```python
# and (E): Apenas retornará Verdade(True) se TODAS as condições forem verdadeiras 
pode_dirigir = (idade >= 18) and possui_cnh
print("Pode dirigir (and):", pode_dirigir)
```

```python
# or (OU): Retorna (Verdade)True se pelo menos UMA condição for verdadeira
e_fim_de_semana = (dia == "Sábado") or (dia == "Domingo")
print("É fim de semana (or):", e_fim_de_semana)
```

```python
# not (NÃO): Inverte o valor lógico
usuario_logado = False
print("Acesso negado (not):", not usuario_logado) 

# Virará True
```

```python
# not (NÃO): Inverte o valor lógico
usuario_logado = True
print("Acesso negado (not):", not usuario_logado) 

# Virará False
```

### 5. Operadores Bitwise (Bit a Bit)

Atuam nos bits individuais de números inteiros em formato binário.

```python
a = 5 
b = 3
```

```python
# & (AND Bitwise): resulta em 1 apenas onde ambos os bits forem 1 (0101 & 0011 = 0001 -> 1)
print("AND bitwise (&):", a & b)
```

```python
# | (OR Bitwise): resulta em 1 se pelo menos um bit for 1 (0101 | 0011 = 0111 -> 7)
print("OR bitwise (|):", a | b)
```

### 6. Operadores Ternário

Permite escrever instruções de condição if/else de forma concisa em uma única linha.

```python
nota = 85
resultado = "Aprovado" if nota >= 60 else "Reprovado"
print("Resultado da nota:", resultado)
```

```python
idade = 18
status = "maior de idade" if idade >= 18 else "menor de idade"
print("Status de idade:", status)
```
