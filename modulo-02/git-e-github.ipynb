# Git & GitHub

### 1. O Problema que o Git Resolve

Quando um projeto evolui sem controle de versão, é comum surgirem pastas com nomes do tipo `portfolio-v1/`, `portfolio-final/`, `portfolio-final-agora-vai/`.   

* Para evitar essa bagunça e a perda de código, o Git resolve as seguintes questões:   
    * O que mudou no código;   
    * Quando mudou;   
    * Quem fez a mudança;   
    * Como comparar ou voltar para uma versão anterior sem depender de cópias manuais.   

### 2. O Problema que o Git Resolve

É fundamental entender que Git e GitHub não são a mesma coisa:

| Git | Github|
|------|-------|
|Sistema de controle de versão local. |Plataforma na nuvem para hospedar repositórios.|
|Funciona localmente na sua máquina, até mesmo sem internet. |Fica disponível na internet para fácil acesso e compartilhamento. |
|Registra o histórico de alterações do seu projeto.|Facilita a colaboração e o gerenciamento de projetos remotos. |


### 3. Preparando o Ambiente Local

Antes de começar a versionar, é necessário identificar a autoria dos seus commits no terminal:

````python
git config --global user.name "Seu Nome"
git config --global user.email "seu.email@exemplo.com"
````

Por que configurar? Os commits registram autoria. Em projetos colaborativos, isso permite saber exatamente quem realizou cada alteração.

### 5. Fluxo Interno do Git: Das Alterações ao Commit

Uma pasta comum não é um repositório Git até ser inicializada com `git init` (que cria a pasta oculta `.git` para guardar o histórico).   
* Quando você trabalha num repositório, o arquivo passa por 3 estados principais:

``` bash
[ Working Directory ]  ---> (git add) --->  [ Staging Area ]  ---> (git commit) --->  [ Histórico (.git) ]
```


* **Working Directory:** Os arquivos em que você está mexendo no momento.   
* **Staging Area:** A área de preparação onde você seleciona o que vai entrar no próximo registro. (Comando: `git add`)   
* **Histórico do Repositório:** O registro definitivo da versão com uma mensagem explicativa. (Comando: `git commit -m "mensagem"`)  

    - ⚠️ Lembre-se: Salvar o arquivo no editor (`Ctrl + S`) apenas salva a alteração no seu disco local. Para salvar uma versão no Git, você precisa fazer o `git add` e o `git commit`. 

### 6. Boas Práticas e Conventional Commits

Escrever boas mensagens de commit ajuda a manter o histórico do projeto legível e rastreável.

|Prefixo|Quando usar|Exemplo|
|-------|-----------|-------|
|`feat:`|Adição de uma nova funcionalidade no sistema.|`git commit -m "feat: adiciona campo de busca de produtos"`|
|`fix:`|Correção de um erro ou bug no código.|`git commit -m "fix: corrige calculo de desconto no carrinho"`|
|`docs:`|Alterações exclusivamente na documentação (como no arquivo `README.md`).|`git commit -m "docs: atualiza instrucoes de instalacao"`|
|`style:`|Ajustes de formatação, espaçamento ou CSS que não alteram a lógica do código.|`git commit -m "style: ajusta alinhamento dos botoes"`|
|`test:`|Adição ou correção de testes automatizados.|`git commit -m "test: adiciona testes unitarios para a API"`|

#### Recomendações:

* Use o verbo no imperativo ou presente indicativo (ex: feat: adiciona botão ao invés de adicionei ou adicionando).   
* Seja direto e objetivo.   
* Evite mensagens vagas como "ajustes" ou "teste".   
* Errou a mensagem do commit antes do push? Corrija com:

```bash
git commit --amend -m "feat: nova mensagem corrigida com sucesso"
```

### 7. Consultando o Histórico e Trabalhando com Repositórios Remotos

Para acompanhar os registros criados e sincronizar com o GitHub:

* Consultar commits no terminal:
    * `git log` — Mostra o histórico detalhado (Autor, data, hash e mensagem).   
    * `git log --oneline` — Exibe o histórico resumido em uma única linha por commit.   
* Integração com o GitHub:
    * `git clone <URL>` — Baixa o repositório remoto do GitHub para o seu computador local.   
    * `git pull` — Puxa as atualizações do repositório remoto no GitHub para o computador local.   
    * `git push` — Envia as alterações commitadas da sua máquina local para o GitHub remoto.
