# Lista de Exercícios: Praticando Git e Github

### Prática 1: Clonando o Repositório da Aula e Navegação no Terminal

**Objetivo:** Baixar o repositório da professora para a máquina local e se familiarizar com os comandos do terminal.   

* Passo a passo:
    * Abra o terminal de sua preferência.   
    * Verifique o seu local atual no terminal executando o comando `pwd`.   
    * Clone o repositório oficial da aula para o seu computador:

``` bash
git clone https://github.com/xavierruth/uc4-logica-de-programacao-python.git 
```

Navegue até a pasta do repositório clonado utilizando `cd NOME-DO-REPOSITORIO` e liste os arquivos presentes com `ls`.   

### Prática 2: Resolvendo o Exercício do Repositório e Aplicando Conventional Commits

**Objetivo:** Alterar um arquivo no repositório clonado, acompanhar o fluxo de versionamento (`add` e `commit`) e utilizar boas práticas de mensagens.   

* Passo a passo:
    * Abra a pasta do projeto no VS Code e acesse o arquivo de exercícios do repositório (ex: `lista-atvds-01.ipynb`).
    * Implemente a resolução de uma das questões propostas.
    * No arquivo `readme.md`, no final, altere o seguinte campo: 
``` bash
        ## 📩 Dúvidas
        * **E-mail:** ruth.xavier@pe.senac.br
        * **LinkedIn:** [linkedin.com/ruthxavier](https://linkedin.com/ruthxavier)
```
para: 
``` bash
        ## Aluno 
        * **Seu nome e sobrenome** 
        * **LinkedIn: [link do seu linkedin](link do seu linkedin)
```
* No terminal, verifique o estado dos arquivos alterados:

    ``` bash 
    git status
    ```
* Adicione as alterações: 

    ```bash 
    git add .
    ```
* Faça o commit aplicando o padrão Conventional Commits e mensagem no imperativo:

    ```bash 
    git commit -m "feat: resolve lista de exercício 1"
    ``` 

* Confira o histórico de versões registradas rodando `git log --oneline`.


### Prática 3: Criando e Personalizando seu README Especial de Perfil

**Objetivo:** Criar um repositório especial no GitHub com o mesmo nome de usuário para personalizar a página inicial do perfil.   

* Passo a passo:
* No GitHub, crie um novo repositório público com o mesmo nome do seu nome de usuário (ex: `usuario/usuario`) e marque a opção `"Add a README file"`.
* Clone este repositório para o seu computador (`git clone + link do respositório a ser clonado`).Abra o arquivo `README.md` e personalize as informações usando a estrutura do modelo abaixo:  

```bash 
# Olá, eu sou o [Seu Nome]! 👋

## 🚀 Sobre mim
- 🎓 Estudando **Técnico em Desenvolvimento de Sistemas - Senac**.
- 💻 Atualmente focado em aprender **[Sua linguagem/foco principal]**.
- 📫 Como me encontrar: [Seu e-mail] ou [Link do LinkedIn].

## 🛠️ Tecnologias e Ferramentas
- **Linguagens:** HTML, CSS, JavaScript, Python
- **Ferramentas:** Git, GitHub, VS Code

## 📊 Estatísticas do GitHub
![Estatísticas do GitHub](https://github-readme-stats.vercel.app/api?username=SEU-USERNAME&show_icons=true&theme=radial)
![Linguagens mais usadas](https://github-readme-stats.vercel.app/api/top-langs/?username=SEU-USERNAME&layout=compact&theme=radial)
```

* Suba as alterações para o repositório executando:

``` bash 
git add README.md
git commit -m "docs: adiciona informacoes e estatisticas ao perfil"
git push origin main 
```
